```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, classification_report
```

```python
categories = ['comp.graphics', 'rec.sport.hockey', 'sci.med', 'talk.politics.mideast']

print("Fetching dataset...")

train = fetch_20newsgroups(subset='train', categories=categories,
                              remove=('headers', 'footers', 'quotes'))

test = fetch_20newsgroups(subset='test', categories=categories,
                              remove=('headers', 'footers', 'quotes'))

X_train, y_train = train.data, train.target
X_test, y_test = test.data, test.target

print("Loaded", len(X_train), "training documents and",
         len(X_test), "test documents.")
```

```python
m = CountVectorizer(stop_words='english', max_features=10000)
X1 = m.fit_transform(X_train)
X2 = m.transform(X_test)

b = CountVectorizer(stop_words='english', max_features=10000, binary=True)
X3 = b.fit_transform(X_train)
X4 = b.transform(X_test)

print("Vectorization strategies initialized and applied.")
```

```python
mnb = MultinomialNB(alpha=1.0)
mnb.fit(X1, y_train)
p1 = mnb.predict(X2)

bnb = BernoulliNB(alpha=1.0)
bnb.fit(X3, y_train)
p2 = bnb.predict(X4)
```

```python
results = {
        'Multinomial NB': [accuracy_score(y_test, p1),
                           f1_score(y_test, p1, average='macro')],
        'Bernoulli NB': [accuracy_score(y_test, p2),
                         f1_score(y_test, p2, average='macro')]
    }

df = pd.DataFrame(results, index=['Accuracy', 'Macro F1-Score'])
print(df.round(4))
```

```python
print("\nMultinomial Naïve Bayes")
print(classification_report(y_test, p1, target_names=categories))

print("\nBernoulli Naïve Bayes")
print(classification_report(y_test, p2, target_names=categories))
```
